```python
import yaml

data_yaml_path = f"{dataset.location}/data.yaml"
with open(data_yaml_path) as f:
    data_cfg = yaml.safe_load(f)
```

```python
print("클래스 목록:", data_cfg["names"])
print("클래스 개수:", data_cfg["nc"])
```

```python
from ultralytics import YOLO

data_yaml_path = f"{dataset.location}/data.yaml"

model = YOLO("yolov8n.pt")

results = model.train(
    data=data_yaml_path,
    epochs=10,
    imgsz=640,
    batch=16,
    project="fall_detect_v2",
    name="yolov8n_2cls"
)
```

```python
best_pt_path = f"{results.save_dir}/weights/best.pt"
print(best_pt_path)
```
